# Average Happiness Score by Region

Average 2021 ladder score for each of the report's 10 regions.

```python
import sys
from pathlib import Path

# Make happiness.py importable whether Jupyter starts in JupyterCodes/ or the repo root
for folder in (Path.cwd(), Path.cwd().parent):
    if (folder / "happiness.py").exists():
        sys.path.insert(0, str(folder))
        break

import matplotlib.pyplot as plt
import pandas as pd
import happiness as hp

df = hp.load_data()
print(f"{len(df)} countries, {df[hp.REGION].nunique()} regions")
```

```python
scores = hp.region_scores(df)
scores.round(2)
```

```python
fig = hp.region_bar_chart(df)
hp.save(fig, "Region score.png")
plt.show()
```

North America & ANZ and Western Europe lead by a wide margin; South Asia and Sub-Saharan Africa have the lowest averages.
